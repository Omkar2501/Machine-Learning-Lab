```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from pandas.core.common import random_state
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv("Salary_Data.csv")
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
#scattered with box for graph
plt.scatter(data['YearsExperience'],data['Salary'])
plt.title('Year Of Experience vs Salary')
plt.xlabel('Year Of Experience')
plt.ylabel('Salary')
plt.box(True)
plt.show()
```

```python
#scattered without box
plt.scatter(data['YearsExperience'],data['Salary'])
plt.title('Year Of Experience vs Salary')
plt.xlabel('Year Of Experience')
plt.ylabel('Salary')
plt.box(False)
plt.show()
```

```python
X=data.iloc[:, :1]
y=data.iloc[:, 1:]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
linearModel=LinearRegression()
linearModel.fit(X_train, y_train)
```

```python
y_pred_test = linearModel.predict(X_test)    
y_pred_train = linearModel.predict(X_train)   
```

```python
plt.scatter(X_train, y_train, color = 'lightcoral')
plt.plot(X_train, y_pred_train, color = 'blue')
plt.title('Salary vs Experience (Training Set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend(['X_train/Pred(y_test)', 'X_train/y_train'], title = 'Sal/Exp', loc='best', facecolor='pink')
plt.show()
```

```python
plt.scatter(X_train, y_train, color = 'lightcoral')
plt.plot(X_train, y_pred_train, color = 'blue')
plt.title('Salary vs Experience (Training Set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend(['X_train/Pred(y_test)', 'X_train/y_train'], title = 'Sal/Exp', loc='best', facecolor='pink')
plt.show()
```

```python
print(f'Coefficient: {linearModel.coef_}')
print(f'Intercept: {linearModel.intercept_}')
```

```python
r_sq1 = linearModel.score(X, y)
print(f"coefficient of determination: {r_sq1}")
```

```python
y_pred1 = linearModel.predict(X)
print(f"predicted response:\n{y_pred1}")
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
```

```python
mean_a = mean_absolute_error(y_test,y_pred_test)
print(mean_a)
```

```python
mean_s = mean_squared_error(y_test,y_pred_test)
print(mean_s)
```

```python
import numpy as np
mean_s_e = mean_squared_error(y_test,y_pred_test)
error=np.sqrt(mean_s_e)
print(error)
```

```python
r_2=r2_score(y_test,y_pred_test)
print(r_2)
```

```python
plt.scatter(X_test, y_test, color="blue", label="Actual")
plt.plot(X_test, y_pred_test, color="red", linewidth=2, label="Predicted Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression: Salary vs Experience")
plt.legend()
plt.show()
```
